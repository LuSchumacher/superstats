## Specify Prior

```python
joint_prior = sup.prior.JointPrior(
    a = sup.transition.GaussianProcess(
        kernel_params={
            "length_scale": sup.prior.Prior("halfnormal", scale=0.3),
            "amplitude": 1.0,
        }
    ),
    b = sup.transition.GaussianProcess(
        kernel=sup.transition.kernel.RBFKernel() + sup.transition.kernel.LinearKernel(),
        kernel_params={
            "length_scale": sup.prior.Prior("halfnormal", scale=0.3),
            "amplitude": 1.0,
            "variance": sup.prior.Prior("halfnormal", scale=2),
        },
        bounds=(0.0, 6.0)
    ),
    c = sup.transition.GaussianProcess(
        kernel=sup.transition.kernel.RBFKernel() * sup.transition.kernel.LinearKernel(),
        kernel_params={
            "length_scale": sup.prior.Prior("halfnormal", scale=0.3),
            "amplitude": 1.0,
            "variance": sup.prior.Prior("halfnormal", scale=0.15),
        }
    ),
    d = sup.transition.GaussianProcess(
        kernel=sup.transition.kernel.LinearKernel(),
        kernel_params={
            "variance": sup.prior.Prior("halfnormal", scale=2),
        }
    ),
)
```

```python
fig = joint_prior.plot_joint_prior(
    num_steps=100,
    num_trajectories=10
)
```

```python
joint_prior = sup.prior.JointPrior(
    v = sup.transition.RandomWalk(
            bounds=(-1.0, 1.0),
            sigma=sup.prior.Prior(dist="halfnormal", scale=0.1),
            delta=sup.prior.Prior(dist="halfnormal", scale=0.02),
            initial_prior=sup.prior.Prior(dist="normal", low=0.0, high=1.0)
        ),
    a = sup.transition.Mixture(
        transitions=[
            sup.transition.AutoRegression(),
            sup.transition.Jump()
        ],
        mixture_weights=sup.prior.Prior(dist="dirichlet", alpha=[25.0, 1.5]),
        bounds=(0.0, 4.0)
    ),
    bias = 0.5,
    tau = sup.prior.Prior(dist="halfnormal", scale=0.15)
)
```

```python
fig = joint_prior.plot_time_varying_prior(
    num_steps=300,
    num_trajectories=10
)
```

```python
fig = joint_prior.plot_time_invariant_prior(num_cols=2)
```

```python
fig = joint_prior.plot_joint_prior()
```

## Specify Generative Model

```python
ddm = sup.simulation.sample_ddm
```

```python
generative_model = sup.simulation.GenerativeModel(
    prior=joint_prior,
    model=ddm,
)
```

```python
plot = generative_model.plot_push_forward(
    num_sim=10,
    num_steps=200,
    data_dim="response_time",
    kind="dist",
    # aggregation=np.median,
    uncertainty_fun="95ci",
    marginal=True,
    # spaghetti=False,
    # num_cols=5,
    # figsize=(15, 6)
)
```

```python
data = generative_model.sample(
    batch_size=10,
    num_steps=100,
)
```

```python
missing = sup.simulation.data_augmentation.RandomMissingProcess(
    p_missing=sup.prior.Prior("halfnormal", scale=0.5),
    missing_value=-100
)
```

```python
aug_data = missing.apply({key: data[key] for key in generative_model.data_keys})
```

## Specify Workflow

```python
workflow = sup.workflow.Workflow(
    simulator=generative_model,
    checkpoint_filepath="basic_ddm"
)
```

### Offline Training

```python
train_data = workflow.simulator.sample(
    batch_size=10,
    num_steps=100,
    tile_to_steps=True
)
test_data = workflow.simulator.sample(
    batch_size=2,
    num_steps=100,
    tile_to_steps=True
)
```

```python
history = workflow.fit_offline(
    data=train_data,
    validation_data=test_data,
    epochs=2,
    batch_size=8
)
```

```python
plot = workflow.plot_history(history)
```

## Online Training

```python
history = workflow.fit_online(
    num_steps=300,
    epochs=75,
    num_batches_per_epoch=500,
    batch_size=32
)
```

```python
plot = workflow.plot_history(history)
```

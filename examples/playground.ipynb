```python
from numba import njit
import numpy as np
from joblib import Parallel, delayed
from functools import wraps
```

```python
@njit
def trial(a, b):
    return a + b

@njit(cache=True, nogil=True)
def trial_loop(a, b):

    x = np.zeros(len(a))
    t = x.shape[0]
    for i in range(t):
        x[i] = trial(a[i], b[i])
    return x
```
